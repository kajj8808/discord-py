```python
from selenium import webdriver
from bs4 import BeautifulSoup

driver = webdriver.Chrome()

redeemcodes = []

driver.get("https://honkai-star-rail.fandom.com/wiki/Redemption_Code")

page_source = driver.page_source

soup = BeautifulSoup(page_source, 'html.parser')

wikitable = soup.select_one('.wikitable')
trs = wikitable.select('tr')


for i, tr in enumerate(trs):
    tds = tr.select("td")
    if len(tds) != 4:
        continue

    def element_get_text(element):
        return element.get_text()

    [code, sever, rewards, duration] = map(element_get_text, tds)
    redeemcodes.append({
        'code': code,
        'server': sever,
        'rewards': rewards,
        'duraion': duration
    })
    print(f"{i}, {code}, {sever}, {rewards}, {duration}")

driver.close()
```

```python
import sqlite3


db = sqlite3.connect("../database/redeemcode.db")
```

```python
drop_table_query = "DROP TABLE Redeemcode;"
db.execute(drop_table_query)
```

```python
create_table_query = """
  CREATE TABLE Redeemcode (
  	id INTEGER PRIMARY KEY AUTOINCREMENT,
    code TEXT UNIQUE,
    server TEXT,
    rewards TEXT,
    duration TEXT
  );
"""

db.execute(create_table_query)
```

```python
insert_query = f"""
    INSERT INTO Redeemcode ('code', 'server', 'rewards', 'duration') VALUES (?,?,?,?);
"""
for redeemcode in redeemcodes:
    def insert_redeemcode():
        code = redeemcode['code']
        exist = db.execute(
            "SELECT * FROM Redeemcode WHERE code = ?;",
            (code,)
        ).fetchone()
        if exist == None:
            # 이 경우에없으니까 여기서 등록 로직 + 알림 기능
            db.execute(
                insert_query, (redeemcode['code'], redeemcode['server'], redeemcode['rewards'], redeemcode['duraion']))
    insert_redeemcode()
```

```python
select_all_query = f"""
    SELECT * FROM Redeemcode;
"""

db.execute(select_all_query).fetchall()
```
